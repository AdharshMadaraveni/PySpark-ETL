# SQL Analytics on IoT Sensor Data

This notebook runs analytical SQL queries on the Delta table.


```python
%sql
SELECT
    sensor_id,
    date_trunc('hour', event_ts) AS hour_window,
    AVG(temperature) AS avg_temperature
FROM gold.iot_sensor_gold
GROUP BY sensor_id, date_trunc('hour', event_ts)
ORDER BY hour_window DESC;
```

```python
%sql
SELECT
    sensor_id,
    COUNT(*) AS overheat_count
FROM gold.iot_sensor_gold
WHERE status = 'Overheat'
GROUP BY sensor_id
ORDER BY overheat_count DESC;
```
